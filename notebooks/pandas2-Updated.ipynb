#### `rename()` method: Renaming columns

Here's what our column names look like in the forest fire dataset:

```python
df.head()
```

- Four of the columns end in `_code`.
- Let's remove that part from the column names.
- We can use the `rename()` method.
- We need to pass the function a **dictionary** of the **old name** to be replaced as the **key** and the **new name** as the **value**.

```python
df.rename(
    columns = {
        "moisture_code": "moisture",
        "fuel_code": "fuel"
        }
        )
```

```python
df.head()
```

Uh-oh, the change didn't stick. DataFrames are **immutable** objects.

We've encountered this before with strings, so we know the answer - **reassign** it to a **variable**.

```python
df = df.rename(
    columns = {
        "moisture_code": "moisture",
        "fuel_code": "fuel"
        }
        )
```

```python
df.head()
```

Alternatively, we can use the `inplace` parameter:

```python
df.rename(
    columns = {
        "moisture_code": "moisture",
        "fuel_code": "fuel"
        },
    inplace=True
    )
```

```python
df.head()
```

### `.drop()` method: Dropping rows and columns

### `.reset_index()` method

- We can **reset** the index and pretend like 2 was never there.
- The `reset_index()` function takes one keyword argument.
- If we don't pass the argument, `drop = True`, an **extra column** will get added to our DataFrame containing the **old index numbers**.

```python
df.head()
```

### Sorting a DataFrame

There are **two methods** for sorting the rows of your DataFrame.

##### `.sort_index()` method

- If you want to sort by the **index numbers**, or if you want to sort by the **column names** (alphabetically), you use `sort_index`.
- It can take two arguments: the axis to sort by (row or column) and the order (ascending or not):

The default arguments are to sort by **row index** with 0 at the top, which is how we've already been viewing the data:

```python
df
```

```python
df.sort_index()
```

Let's try more arguments:

```python
df.sort_index(ascending = False) # Sort index in descending order
```

```python
df.sort_index(axis = 1) # Sort columns in alphabetical order, axis = 0 means we are sorting rows, axis = 1 means we are sorting columns
```

```python
df.sort_index(axis = 1, ascending = False) # Sort columns in reverse alphabetical order
```

##### `.sort_values()` method

The second sort function, `sort_values()`, will sort the rows of the DataFrame by the **data** in a **column**:

```python
df.sort_values("area_burned") # Sort values by the "area_burned" column, by default it sorts in ascending order
```

```python
df.sort_values("day") # Sort values by the "day" column, by default it sorts in ascending order
```

### `.groupby()` method

Often, you will want to calculate the statistics for a **particular subgroup** of a data column.

- For example, let's say we want to ask if more fires happen on certain days of the week.
- This code will tell you the **count** for **every column** in the DataFrame **except** the column that you are using to **group** your data by.

```python
# Group the DataFrame by the "day" column and count how many non-missing values each other column has within each group
# Alternatively: "For each unique day, count how many non-missing entries exist in each column"
df.groupby("day").count()
```

- We can make this easier to read by sorting the rows from lowest count to highest count.
- We can chain multiple method functions onto our code, as long as they go in the correct logical order.
- We will pick any column to sort on, since they are all the same.

```python
# Group the DataFrame by the "day" column, count how many non-missing values each other column has within each group
# and sort the resulting DataFrame by the "month" column in ascending order
df.groupby("day").count().sort_values("month")
```

<br>It looks like weekends have more fires than weekdays.

## <br><br><br>Slightly more complicated examples

### <br>Removing Outliers

Let's look at the correlation between humidity and area_burned:

```python
df.plot.scatter(x = "humidity", y = "area_burned")
```

<br>We can see that a few outliers are clouding any relationship. We can remove them with a boolean, but we need to look at the plot above and decide where to make the cutoff. Let's try getting rid of only points above 400.

```python
df_no_outliers = df[df["area_burned"] < 400]
```

```python
df_no_outliers.plot.scatter(x = "humidity", y = "area_burned")
```

<br><br>There are a large number of points with 0 for area_burned. Let's remove those, and only look at days where the fires spread.

```python
df_no_outliers = df[(df["area_burned"] > 0) & (df["area_burned"] < 400)]
```

```python
df_no_outliers.plot.scatter(x = "humidity", y = "area_burned")
```

### <br><br><br>Plotting with categorical data

Let's say we want to see the relationship between month and humidity. We can try a scatter plot:

```python
df.plot.scatter(x = "month", y = "humidity")
```

<br>That doesn't work! Let's try another type of plot - a bar plot:

```python
df.plot.bar(x = "month", y = "humidity")
```

<br>Also not what we're looking for. It's plotting each data point individually.

What we really want to see is how the mean humidity changes each month. We can create a new DataFrame that only includes the data we need. We will group by month, select only the humidity column, and then find the mean.

```python
hum_mean = df.groupby("month")[["humidity"]].mean()
hum_mean
```

<br>Now we have a nice series object that we can plot:

```python
hum_mean.plot.bar()
```

<br>This is looking better, but we need to sort them by month, not alphabetically. We're going to worry about that in just a minute.

<br><br><br>Let's deal with the sorting issue! We can create a new column in our DataFrame that contains a numerical value for the month. We will use a handy pandas function called `replace()`.
