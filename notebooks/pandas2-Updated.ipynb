### `.groupby()` method

Often, you will want to calculate the statistics for a **particular subgroup** of a data column.

- For example, let's say we want to ask if more fires happen on certain days of the week.
- This code will tell you the **count** for **every column** in the DataFrame **except** the column that you are using to **group** your data by.

````python
# Group the DataFrame by the "day" column and count how many non-missing values each other column has within each group
# Alternatively: "For each unique day, count how many non-missing entries exist in each column"
df.groupby("day").count()
````

- We can make this easier to read by sorting the rows from lowest count to highest count.
- We can **chain multiple methods** onto our code, as long as they go in the **correct logical order**.
- We will pick any column to sort on, since they are all the same.

````python
df.groupby("day").count().sort_values("month") # We can pick any column to sort by, since the values are all the same
````

<br>It looks like weekends have more fires than weekdays.

Note that just grouping the DataFrame does not give us a number. It gives a `GroupBy object` that we need to convert to a number by chaining a method.

````python
df.groupby("day")
````

````python
df.groupby("day").count()
````

### Chaining grouping and indexing

If you only want to see an aggregate total for **one column** in the DataFrame, you can chain column indexing (bracket notation) or `.loc[]`.

````python
df.drop("day", axis = 1).groupby("month").mean().loc[:, ["area_burned"]]
````

Alternatively, with column indexing:

````python
df.drop("day", axis = 1).groupby("month").mean()[["area_burned"]]
````

What is the mean area burned on each day of the week?

````python
# We select the "area_burned" column before applying the .mean() method, as applying .mean() on string columns will give an error.
df.groupby("day")[["area_burned"]].mean()
````

We used column indexing directly on the GroupBy object to select the `area_burned` column. We will come back to that.

<br>Again, we can also sort the data:

````python
df.groupby("day")[["area_burned"]].mean().sort_values("area_burned")
````

<br>So Saturday fires are the most destructive fires.

<br>We can also chain some other methods to our code, like `.round()`:

````python
df.groupby("day")[["area_burned"]].mean().round(2).sort_values("area_burned")
````

#### `.loc[]` and GroupBy objects

In the previous example, we used column indexing and not `.loc[]`:

````python
df.groupby("day")[["area_burned"]].mean()
````

Could we have used `.loc[]`?

````python
try:
    df.groupby("day").loc[:, "area_burned"].mean() # Get the mean of the "area_burned" column for each unique day
except AttributeError:
    print("You cannot use .loc on a GroupBy object. You can only use .loc on a DataFrame or Series.")
````

When we used `.loc[]` earlier, if was not directly chained to the GroupBy object but to the `.mean()` method:

````python
df.drop("day", axis = 1).groupby("month").mean().loc[:, ["area_burned"]] # The order here is different.
````

Similarly, if we wanted to write:

```python
df.groupby("day")[["area_burned"]].mean()
```

using `.loc[]`, we could write:

````python
# We have to drop the "month" column first, which contains strings, to calculate the mean. Otherwise we will get an error.
df.drop("month", axis = 1).groupby("day").mean().loc[:, "area_burned"]
````

However, **column indexing can be chained directly to GroupBy objects**:

````python
df.drop("day", axis = 1).groupby("month")[["area_burned"]].mean()
````

This is one area where column indexing has an advantage over using `.loc[]`.
